##### BASICS OF DSA

```python
# -- Recursion -- 
def countdown(n):
    if n == 0:
        return
    countdown(n-1)
    print(n)

countdown(3)
#O(n)
```

```python
# --- Find Fib Number ---

# 1. Find nth fib number 
def n_fib(n):
    if n<=1:
        return n
    return n_fib(n-1) + n_fib(n-2)
#O(n^2)

# 2. Find till Nth fib number
def fib_n(n):
    result = []
    a,b = 0,1
    for i in range(n):
        result.append(a)
        a,b = b,a+b
    return result

print(fib_n(9))
#O(n)
```

```python
# 3. Find if the n^th value is power of 2 or not


def power_(n):
    while n%2==0:
        n=n//2
    return n==1
# O(log n)

def power_(n):
    if n<=0: return False
    if n==1: return True
    if n%2!=0: return False
    return power_(n//2)
# O(log n)
```

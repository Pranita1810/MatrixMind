```python
# --- Find Largest Value ---
lst = [1,666,775,33,6,2,8,99]
largest = 0
for i in lst:
    if i > largest:
        largest = i
print(largest)

# --- Find Second Largest Value ---
largest = 0
sec_lar = 0
for i in lst:
    if i > largest:
        sec_lar = largest
        largest = i
    elif i > sec_lar and i!=largest:
        sec_lar = i
print(sec_lar)
# -- O(n) --
```

```python
# --- Check if the arrya is sorted ---
lst = [66,1,2,3,4,5,6,99]

sorted_ = None
for i in range(len(lst)-1):
    if lst[i] < lst[i+1]:
        sorted_=True
    else:
        sorted_=False
        break

print(sorted_)


# --- Remove Duplicate from sorted array ---
# 1. Brut Force
lst = [1,1,2,3,4,4,5,5]
em_set = set()
for i in lst:
    em_set.add(i)
print(em_set)

# 2. DSA
j = 0
for i in range(len(lst)-1):
    if lst[i] != lst[j]:
        j = j+1
        lst[j] = lst[i]

print(lst[:j+1])
```

```python
# --- Rotate the array from left ---
array = [1,2,3,4,5]
temp = array[0]
for i in range(len(array)-1):
    array[i] = array[i+1]
array[-1] = temp
print(array)


# --- Rotate the arrya from K ---
arr = [1,2,3,4,5]
k=3
if k < len(arr):
    arr[k:] = reversed(arr[k:])
    arr[:k] = reversed(arr[:k])
    arr.reverse()
print(arr)
```

##### Regular Expression
- Regular Expression or RegEx is simply text search engine that searches the value based on it's value, pattern.

```python
# --- RegEx ---
import re

text = "my name is pranit and my phone number is 8850559967/885055996777/88505599671"
phone = re.findall(r"\b\d{10}\b", text)    # -> Find exactly 10 numbers (if not using \b it i will return 10 number even from 11 or futher sequences of numbers)
searching = re.search(r"\d+", text)        # -> Search will only return first result.
word = re.findall(r"\w", text)             # -> only \w will return list containing each letter
words = re.findall(r"\w+",text)            # -> This will return all vlues in list type.
o_wods = re.findall(r"\D+",text)           # -> This will return only string not int
r_mat = re.sub(r"\d+", "xxxxxxxxxxxx", text) # -> Replace matched value

mail_text = "his full email is pranitugale1000@gmail.com"
mail_o = re.findall(r"\w+@\w+\.\w+", mail_text)
print(mail_o)
```

##### ITERATORS & GENERATORS
- These are related to each other which helps to interact and perform operations on sequence data.
- Iterators is an object that alllows user to access all elements.
- Genearator are simply python obhects that also helps to access elements but not in one go.

```python
# -- Iteration --  
numbers = [1,2,3]       # -> Auto Iteration
for i in numbers:
    print(i)

iterator = iter(numbers) # -> These are manual iteration
print(next(iterator))
print(next(iterator))
print(next(iterator))


# -> The main difference between iterators & generators is iterator iter on existing sequenc data where as generator create new data when we call it.


# -- Generator --
def generate():
    yield 10
    yield 20
    yield 30
    yield 70

gen = generate()
val_10 = next(gen)
val_20 = next(gen)
val_30 = next(gen)


# ------ REAL LIFE USAGE -----
# It is mainly used for lazy loading & lazy processing.
# Example 1. : we have 10M rows and want to process load only 1000 + multiprocessing to perform operations.
```

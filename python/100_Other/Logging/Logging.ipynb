##### LOGGING
- Logging means keeping record of what a program is doing. It's like keeping record of 
  what and when happened.
  

```python
# Instead of writing: 
print("Database connected successfully")
print("ETL Process started")
print("Data stored successfully")

# we use:
import logging

logging.info("Database connected successfully")
logging.error("Database connection failed")
logging.debug("Debuggin problems")
logging.warning("Failed! try again")


# --- Real World Setup ---
import logging

logging.basicConfig(
    filename="app.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)

logging.info("ETL started")
logging.info("Extracted 10,000 rows")
logging.info("Transformation completed")
logging.error("Database connection failed")
```

```python
# 1. 4x4 Start pattern
for i in range(4):
    for j in range(4):
        print("*", end="")
    print()
```

```python
# 2 Print inremental stars
for i in range(5):
    print("*"*i)

# 3 Print incremental numbers
for i in range(1,6):
    for j in range(1, i + 1):
        print(j, end="")
    print()

# 4. Print same number increase by n
for i in range(1,6):
    for j in range(1, i + 1):
        print(i, end="")
    print()
```

```python
# 5. Print decreasing starts
x =  5
for i in range(5):
    print("*"*x)
    x = x - 1

#6. Print decresing numbers
x = 6
for i in range(1,6):
    for j in range(1,x):
        print(j, end='')
    print()
    x = x -1
```

```python
# 7. Print a triangle
for i in range (5):
    for j in range(4-i):
        print(" ", end="")
    for j in range(i):
        print("* ", end="")
    print()

for k in range(5):
    print(" "*(5-k)+"* " * k )

for k in range(5):
    print(" "*(k)+"* " *(5-k) )
```

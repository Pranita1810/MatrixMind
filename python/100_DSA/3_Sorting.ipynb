```python
# Bubble sorting an array

arr = [7,4,2,6,9,5,3]

for j in arr:
    for i in range(len(arr)-1):
        if arr[i] > arr[i+1]:
            temp = arr[i+1]
            arr[i+1] = arr[i]
            arr[i] = temp

print(arr)
```

```python
# Insertion sorting

arr = [4,6,8,3,2,5,8]

for i in range(1, len(arr)):
    key = arr[i]
    j = i -1
    while j>=0 and arr[j]>key:
        arr[j+1] = arr[j]
        j = j - 1
    arr[j+1] = key

print(arr)

```

```python
# Selection Sorting

arr = [64, 25, 12, 22, 11]

for i in range(len(arr)):
    min_ = i
    for j in range(i+1, len(arr)):
        if arr[j] < arr[min_]:
            min_ = j

    arr[i], arr[min_] = arr[min_], arr[i]

print(arr)
```

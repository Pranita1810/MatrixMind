```python
# --- LINEAR SEARCH ---
def lin_search(arr, target):
    if target not in arr:
        return -1
    for i in range(len(arr)):
        if arr[i] == target:
            return i
```

```python
# --- LARGEST ELEMENT ---
def largest(arr):
    lar = 0
    for i in arr:
        if i > lar:
            lar = i
    return lar

# --- SECOND LARGEST ---
def sec_largest(arr):
    lar = 0
    sec_lar = 0
    for i in arr:
        if i > lar:
            sec_lar = lar
            lar = i
        elif i>sec_lar and sec_lar!=lar:
            sec_lar = i
    return sec_lar
```

```python

arr = [1,0,1,0,1,1,0,0,0]

def consecutive(arr):
    count = 0
    max_count = 0

    for i in range(len(arr)):
        if arr[i] == 1:
            count += 1
            max_count = max(count, max_count)
        else:
            count = 0

    return max_count

print(consecutive(arr))
```

```python
arr = [1,2,3,4,5,6]

# Left rotate arrya by one
temp = arr[0]
arr.remove(arr[0])
arr.append(temp)
    

# Rotate an array from K
k = 2
temp = []
for i in range(len(arr)):
    if i<k:
        temp.append(arr[i])
arr = arr[k:]+temp
```

##### MULTI THREADING
- Multithreading is when python code uses multiple thread & single CPU core to perform any set of actions. Multithreading is very useful for I/O operations because it uses concurrency not true multithreading.
- For e.g. - A user want to send 5 requests to differnt APIs assuming each API takes around 30-40 sec to respondse. Without multithreading : Execute one by one 5x40 = 200 seconds in total. Using multithreading : Send first request, while waiting for first execute second & so on 

```python
import threading
import time

def task(name):
    print(name, "Start")
    time.sleep(5)
    print(name, "Done")

t1 = threading.Thread(target=task, args=("Task 1",))
t2 = threading.Thread(target=task, args=("Task 2",))
t3 = threading.Thread(target=task, args=("Task 3",))

t1.start()
t2.start()
t3.start()

t1.join()
t2.join()
t3.join()

print("All tasks executed successfully")

#----------------------------------------------------------------------------------------------------------------------------------

#Instead of manually creating 10 threads, python manages them for you
from concurrent.futures import ThreadPoolExecutor
import time

def download(file):
    print("Downloading", file)
    time.sleep(3)
    return f"{file} Downloaded"


with ThreadPoolExecutor(max_workers=3) as executor:
    result = executor.map(download,["file1", "file2" , "file3"])
    for i in result:
        print(i)

with ThreadPoolExecutor(max_workers=5) as submit:
    respondse_1 = submit.submit(download, "path")
    respondse_2 = submit.submit(download, "path2")
    respondse_3 = submit.submit(download, "path3")

    
```

##### MULTI PROCESSING
- Multiprocessing is a used for CPU bound tasks. CPU tasks are operations where it's speed limit is based on CPU specification rather than external source like ssd, ram or hdd. 
- It is great to use for maths, cpu heavy work, ML.

```python
from concurrent.futures import ProcessPoolExecutor

def calculate(n):
    return sum(i * i for i in range(n))

if __name__ == "__main__":
    with ProcessPoolExecutor(max_workers=3) as executor:
        results = list(executor.map(calculate, [1_000_000, 2_000_000, 3_000_000]))

    print(results)


from concurrent.futures import ProcessPoolExecutor

def calculate(n):
    return n * n

if __name__ == "__main__":
    with ProcessPoolExecutor(max_workers=3) as executor:
        f1 = executor.submit(calculate, 10)
        f2 = executor.submit(calculate, 20)
        f3 = executor.submit(calculate, 30)

        results = [f1.result(), f2.result(), f3.result()]

    print(results)  # [100, 400, 900]
```

##### ASYNC
- Async let a single thread handle multiple I/O operations without blocking while waiting. Which is specially used for building APIs that communicate with databases.

```python
import asyncio

async def fetch_data(name):
    print(f"{name} started")
    await asyncio.sleep(2)  # Simulates non-blocking I/O
    print(f"{name} completed")

async def main():
    await asyncio.gather(
        fetch_data("API 1"),
        fetch_data("API 2"),
        fetch_data("API 3")
    )

asyncio.run(main())
```

##### request
- It is a basically HTTP client
- Python Code -> requst -> HTTP Client -> server -> respondse -> Python
- Header : It is an extra information you send along with HTTP requst to tell server about your request/client.

```python
import requests

# -- HTTP simple request --
url = "https://www.geeksforgeeks.org/python/python-web-scraping-tutorial/"
respondse = requests.get(url)         
print(respondse.text)

# -- request with header --
requests.get(url, headers={"User-Agent": "Mozilla/5.0"})         # -> Pretend to be a browser 
requests.get(url, headers={"Accept": "application/json"})        # -> Ask for JSON
requests.get(url, headers={"Authorization": "Bearer YOUR_TOKEN"})# -> Send API authentication
```

#### bs4
- bs4 is a HTML parser that read the HTML code and convert it into human readable format, basically it extract the text we need from defined elements like 
  h1, h2, h3, tables etc.

```python
from bs4 import BeautifulSoup

# -- Get HTTP Content using request --
url = "https://www.geeksforgeeks.org/python/python-web-scraping-tutorial/"
respondse = requests.get(url).text    

html_content = BeautifulSoup(respondse, "html.parser")         # -> Convert raw html content in structured one
for i in html_content.find_all("h2"):                          # -> Find each h2 content 
    h2_content = i.text


# ---- EXAMPLE PROJECT ----
url = "https://housing.com/in/buy/mumbai/"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/151.0.0.0 Safari/537.36",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8",
    "Accept-Language": "en-IN,en;q=0.9",
    "Accept-Encoding": "gzip, deflate, br, zstd",
    "Referer": "https://housing.com/",
    "Origin": "https://housing.com",
    "Sec-Fetch-Dest": "document",
    "Sec-Fetch-Mode": "navigate",
    "Sec-Fetch-Site": "same-origin",
    "Sec-Fetch-User": "?1",
    "sec-ch-ua": '"Chromium";v="151", "Google Chrome";v="151", "Not_A Brand";v="99"',
    "sec-ch-ua-mobile": "?0",
    "sec-ch-ua-platform": '"Windows"',
    "Upgrade-Insecure-Requests": "1",
}



respondse = requests.get(url=url, headers=headers)
parsing = BeautifulSoup(respondse, "html.parser")
print(parsing.text)
for i in parsing.find_all("h2").text:
    print(i.text)
```

##### SELENIUM
- Selenium controls web browser like human, it is not backend process it practically opens the browser and act like human.
- It's like automated mouse and keyboard that works on web browsers.

NOTE : selenium is all about baby steps building.

```python
# --- Basic Usage ---
from selenium import webdriver                                   
#Web-Driver
driver = webdriver.Chrome()                                         # -> Define webdriver  
driver.get("https://www.selenium.dev/documentation/webdriver/")     # -> Send HTTP request to server and get results.


#Find Element & automate basic tasks
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
driver = webdriver.Chrome()
driver.get("https://www.selenium.dev/documentation/webdriver/")                 #-> Open the URL.
search_box = driver.find_element(By.CLASS_NAME, "DocSearch-Button-Placeholder") #-> Find search box by key.
search_box.click()                                                              #-> click on that search box.
search = driver.find_element(By.ID, "docsearch-input")                          #-> Go to the search box. 
search.send_keys("search")                                                      #-> Search word "anything" and enter.
search = driver.find_element(By.ID, "docsearch-input")               
search.send_keys(Keys.ENTER)
```

##### PYTHON SCHEDULING
- Python scheduling is simply running an individual or multiple programs/softwares or processes automatically on given time.

```python
import schedule
import time

def job1():
    print("extract")

def job2():
    print("transform")

def job3():
    print("load")

schedule.every(10).seconds.do(job1)
schedule.every().day.at("10:39").do(job_func=job2)
schedule.every().friday.at("10:39").do(job_func=job3)

schedule.run_all() #-> This is used to run all jobs at once

while True:
    schedule.run_pending()
    time.sleep(10)
```

##### ADVANCE PYTHON SCHEDULER
- It is Prod level scheduler just like schedule module but on steroids.

```python
from apscheduler.schedulers.background import BackgroundScheduler
from datetime import datetime

def job1():
    print("Extract")

def job2(msg):
    print("Transform")
    print(msg)

def job3():
    print("Load")

scheduler = BackgroundScheduler()
scheduler.add_job(job1, "interval", seconds=5)
scheduler.add_job(job2,"cron",day_of_week="fri",hour=11,minute=46, args="message")
scheduler.add_job(job3,"date",run_date=datetime(2026, 9, 4, 15, 30))
scheduler.start()
```

##### RDBMS/DBMS Using PYODBC
- There are several libraries & modules like pyodbc, sqlite, Postgresql, SQLAlchemy. we will learn pyodbc and SQLAlchemy because these ae most used in industry.

```python
# ---------- 1. SSMS ----------
import pyodbc
import pandas as pd

class SQLServerDB:
    def __init__(self):
        self.server = r"localhost\SQLExpress"
        self.driver =   "{OODBC Driver 17 for SQL Server}"
        self.database = "test"

        self.conn = pyodbc.connect(
            f"SERVER={self.server};"
            f"DRIVER={self.driver};"
            f"DATABASE={self.database};"
            "Trusted_Connection=yes;"
        )

    def read_data(self, query):
        return pd.read_sql(query, self.conn)

    def execute_query(self, query):
        cursor = self.conn.cursor()

        try:
            cursor.execute(query)
            self.conn.commit()

        except Exception:
            self.conn.rollback()
            raise

        finally:
            cursor.close()

    def close(self):
        self.conn.close()



# ---------- 2. SQLlite ----------
import pyodbc
import pandas as pd

class SQLite_op:
    def __init__(self):
        self.DRIVER = "{SQLite3 ODBC Driver}"
        self.DATABASE = "test.db"
        self.conn = pyodbc.connect(f"DRIVER={self.DRIVER};" f"DATABASE={self.DATABASE};")

    def read_data(self, query):
        return pd.read_sql(query, self.conn)

    def _execution_query(self, query2):
        with self.conn.cursor() as conn_cur:
            conn_cur.execute(query2)
            conn_cur.commit()

# ---------- 3. POSTgres ----------
import pyodbc
import pandas as pd

class PostGre:
    def __init__(self):
        self.SERVER = "127.0.0.1"
        self.DATABASE = "postgres"
        self.UID = "postgres"
        self.PW = "pranit"
        self.DRIVER = "{PostgreSQL Unicode}"
        self.PORT = 5432

        self.conn = pyodbc.connect(f"DRIVER={self.DRIVER};" f"SERVER={self.SERVER};" f"PORT={self.PORT};" f"UID={self.UID};"
                                   f"PWD={self.PW};" f"DATABASE={self.DATABASE}")

    def _read_data(self, query):
        return pd.read_sql(query, self.conn)

    def _execution(self, query2):
        with self.conn as con:
            con.execute(query2)


# -------- 4. MySQL ----------
import pyodbc 
import pandas as pd

class My_SQL_op:
    def __init__(self):
        self.DRIVER = "{MySQL ODBC 26.7 Unicode Driver}"
        self.SERVER = "localhost"
        self.UID = "root"
        self.PWD = "pranit"
        self.DATABSE = "mysql"

        self.conn = pyodbc.connect(f"DRIVER={self.DRIVER};" f"SERVER={self.SERVER};" f"UID={self.UID};" 
                                   f"PWD={self.PWD};" f"DATABASE={self.DATABSE};")

    def read_data(self, query):
        return pd.read_sql(query, self.conn)

    def _execution_(self, query2):
        with self.conn.cursor() as conn:
            conn.execute(query2)
        self.conn.commit()

        
```

##### RDBMS Using Industry Standard Modules

```python
# --------- SSMS ---------
import sqlalchemy
import pandas as pd

class SQLOperation:
    def __init__(self):
        self.server = r"localhost\SQLExpress"
        self.database = "test"
        self.driver = "ODBC Driver 17 for SQL Server"

        self.engine = sqlalchemy.create_engine(
            f"mssql+pyodbc://@{self.server}/{self.database}"
            f"?driver={self.driver.replace(' ', '+')}"
            "&trusted_connection=yes"
        )

    def read_data(self, query):
        return pd.read_sql(query, self.engine)

    def execute_query(self, query2):
        try:
            with self.engine.begin() as conn:
                conn.execute(sqlalchemy.text(query2))
        except Exception as e:
            return e



# ---------- 2. SQLlite ----------
import sqlite3
import pandas as pd

class SQLite:
    def __init__(self):
        self.conn = sqlite3.connect("test.db")
        self.cursor = self.conn.cursor()

    def read_df(self, query):
        return pd.read_sql(query, self.conn)

    def execution(self, query):
        with self.conn as conn:
            conn.execute(query)



# ------------ 3. MYSql -----------
import mysql.connector
import pandas as pd

class MySQL_op:
    def __init__(self):
        self.HOST =  "localhost"
        self.DATABASE = "mysql"
        self.UID = "root"
        self.PWD = "pranit"
        self.conn = mysql.connector.connect(host=self.HOST, database=self.DATABASE, user=self.UID, password=self.PWD)

    def read_data(self, query):
        return pd.read_sql(query, self.conn)

    def _execute_query(self, query2):
        with self.conn.cursor() as cursor:
            cursor.execute(query2)
        self.conn.commit()



# --------- 4. PostGre ---------
import psycopg2
import pandas as pd

class PostGre_operations:
    def __init__(self):
        self.HOST = "localhost"
        self.DATABASE = "postgres"
        self.UID = "postgres"
        self.PWD = "pranit"
        self.PORT = 5432

        self.conn = psycopg2.connect(host=self.HOST, database = self.DATABASE, user=self.UID, password=self.PWD, port=self.PORT)

    def _execute(self, query):
        with self.conn.cursor() as cursor:
            cursor.execute(query)
        cursor.commit()
```

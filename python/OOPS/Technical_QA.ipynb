```python
# CONSTRUCTOR - __init__ & MEMORY MANAGEMENT
class company:
    def __init__(self):             #-> __init__ is a constructor that calls itself when an instance of an objected is created.
        self.salary = 25000
        self.id = 10
        
    @staticmethod
    def temp():
        age = 22
        return age
        
s = company()
print(s.__dict__)         # -> This __dict__ is a magic method used to view content of the class attributes. | Here we also accessed the attribute of an Object
print(s.temp())           # -> if s.temp is not staticmethod or you did not write self in temp(self) as parameter it will give an error. | Here we accessed the method of that object
```

```python
# BUILT IN FUNCTIONS & BUILT IN CLASS ATTRIBUTES 
class Employee:
    '''This is used to maintain the documentation of the class'''
    def __init__(self, name, salary, pos):
        self.name = name
        self.salary = salary
        self.pos = pos
    
    def get_info(self):
        print(self.name, self.salary, self.pos)
        
s = Employee("pranit", 100000, "DS")
s.get_info()


# 1) getattr 2)setattr 3)delattr 4)hasattr - These built in functions are used to manage attributes in the class
setattr(s,"name","siddhi")
getattr(s,"name")
delattr(s,"name")
hasattr(s, "salary")


# 1) __dict__   2)__doc__   3)__name__   4)__module__   5)__bases__
print(Employee.__dict__)     #-> returns the attributes and values in dict data type
print(Employee.__doc__)      #-> Return comment 
print(Employee.__name__)     #-> Return name of the clas
print(Employee.__module__)   #-> Used in inheritance to return the module from where it came.
```

```python
#INSTANCE VS CLASS METHODS & VARIABLES
class Student:
    college = "ABC"          # class variable

    def __init__(self, name, marks):
        self.name = name     # instance variable
        self.marks = marks   # instance variable

    def show(self):          # instance method
        print(self.name, self.marks, Student.college)
        

    @classmethod
    def change_college(cls, cname):   # class method
        cls.college = cname

    @staticmethod
    def is_pass(marks):      # static method
        return marks >= 40


s1 = Student("Pranit", 55)
s2 = Student("Siddhi", 35)

s1.show()
s2.show()
Student.change_college("XYZ")
s1.show()
print(Student.is_pass(55))
```

```python
#INHERITANCE 
# There are mainly 3 types of inheritance that is single, mulit level & multi
# 1 ----- SINGLE -----
class dad:
    def __init__(self, face, hair, voice):
        self.face = face
        self.hair = hair
        self.voice = voice 
    def type_gen(self):
        print(f"His voice is {self.voice} and face is {self.face} and hairs are {self.hair}")
class son(dad):
    def both():
        dad.type_gen()
x = son("Dimond", "Black", "High")
x.type_gen()


#2. ----- HIERARCHICAL ------
class Grandfather:
    def __init__(self, color, nature):
        self.color = color
        self.nature = nature
        print("But the main triat is crazy")

class fahter(Grandfather):
    def __init__(self, color, nature):
        super().__init__(color, nature)
        print(color, nature)
    
class daughter(fahter):
    def __init__(self,color, nature):
        super().__init__(color, nature)
        print("Nothing")

s = daughter("Blond", "Kind")


# 2 ----- MULTI INHERITANCE -----
class fahter(Grandfather):
    def __init__(self, color, nature):
        super().__init__(color, nature)

class mom(Grandfather):
    def __init__(self, color, nature):
        super().__init__(color, nature)

class common_son(fahter, mom):
    def __init__(self, color, nature):
        super().__init__(color, nature)
        print(color, nature)
b = common_son("Black", "Aesthetic")
```

```python
#CONSTRUCTOR OVER-RIDING
# Constructor over riding is when child class defines it's own constructor which over rides the parent's constructor
class manager:
    def __init__(self,x):
        self.temp = "No work"
        print(self.temp)

class employee(manager):            
    def __init__(self, y):
        self.temp2 = "Hard work"
        print(self.temp2)
    
s = employee()     
# so for only one instance only one constructor can be used, if we want to use both constructors we need to use __super__ method

#--------------------------------------------------------------------------------------------------------------------------------------------------------------------------

#SUPER METHOD
class employee(manager):
    def __init__(self,x):
        super().__init__(x)              # Child constructor still overrides the parent constructor, but super() explicitly calls the parent constructor.
        print("Over Riding constructor")
    
s = employee("x")
 
# Note: if any constructor has parameter we have to pass same parameter's variable in super().__init__(x,y parameter) and in constructor of child class
```

```python
# # ENCAPSULATION
# # Building attributes and mthods inside the class and controlling their access to the outside world is what defines encapsulation.
class account_details:
    def __init__(self):
        self.__id = "A1456"                 # By adding double underscore it make it privet which cannot be used outside of the class
        self.__pass = "niga"

    def display(self):
        dis_id = self.__id
        dis_pas = self.__pass
        return dis_id, dis_pas  
    
s = account_details()
print(s._account_details__id)         # But here they are still acceseble when you apply double underscore python just change the name of that attribute. from __id to _classname__id.
print(s.display())
```

```python
#POLYMORPHISM
# It means many forms, it's like one thing has differnt nature or behaviour. Polymorphism in python is an ability of python object to take many forms. For e.g. 2+2 is 4 but a+b is ab according to the input it changed it's behaviour
lst = [1,2,3,4,5]
string = "pranit"

print(len(lst))         #-> Changed it's behaviour for list
print(len(string))      #-> Chnaged it's behaviour for string


#POLYMORPHISM + INHERITANCE
class bike:
    def __init__(self, type_, name):
        self.type = type_
        self.name = name

    def return_dets(self):
        print(self.type)
        print(self.name)

    def max_speed(self):
        print("Max speed is 100")
    
    def gear(slef):
        print("Max gears are 4")

class car(bike):
    def max_speed(self):
        print("Max speed is 200")
    
    def gear(slef):
        print("Max gears are 6")

b = bike("2 wheels", "Bajaj")
b.return_dets()
b.max_speed()
b.gear()

c = car("4 Wheels", "Tata")
c.return_dets()
c.max_speed()
c.gear()
```

```python
#INNER CLASS
class University:
    def __init__(self, name, owner):
        self.name = name
        self.owner = owner

    def display_uni_info(self):
        print(f"Name : {self.name} , Owner : {self.owner}")

    class School:
        def __init__(self, name, trustee):
            self.name = name
            self.trustee = trustee 
        
        def display_scho_info(self):
            print(f"Name : {self.name} , Trustee : {self.trustee}")

u=University("Mumbai", "Shinde")
u.display_uni_info()
s = u.School("Tolani", "shinde")
s.display_scho_info()
```

```python
# DECORATORS
# In OOPS decorator is when you chnage the output or methodology of an existing function.
def prniter():
    print("Hi") 
    print("Hi")              #-> Created a normal function that prints Hi two times

def decor(func):             #-> Created a decorator function that behaves same as printer function (now we will chnage the behaviour in next example).
    func()                   #-> Calling function inside the deco funtion
    
decor(prniter)               #-> Finally calling deco funtion which calls printer automatically cuz it is inside decor function. 


#--- Example 2 ---
def decor_(printer_):
    def inner():
        printer_()
        print("Hello from decor")
    return inner

@decor_
def printer_():
    print("Hello")
    print("Hello")

printer_()


# --- Example 3 ---
def substraction(addition):
    def inner():
        op_input = input("Enter Operation type")
        if op_input == "A":
            result = addition()
            return result
        if op_input == "S":
            usr_input1 = int(input("Enter num here"))
            usr_input2 = int(input("Enter num here"))
            result_sub = usr_input1 - usr_input2
            return result_sub
    return inner

@substraction
def addition():
    usr_input1 = int(input("Enter num here"))
    usr_input2 = int(input("Enter num here"))
    results = usr_input1 + usr_input2
    return results

print(addition())

```

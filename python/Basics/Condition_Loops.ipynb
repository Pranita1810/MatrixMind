##### MATCH / CASE
In python 3.10 version it is a powerfull mechnism for pattern matching. Unlike if else statement whihc can become messy with complex logic
the match case provides more elegent and flexible solution.

```python
# Use it when the data is structured or checking conditions for single variable
# -- Use when single variable condition matching
x = 1 
match x:
    case 1:
        print("It is 1")
    case 2:
        print("It is 2")
    case _:
        print("Nothing")


# -- Use when data is structured
data = ("error", 500)
match data:
    case ("error", code):
        print("Error:", code)
    case ("success", code):
        print("Success:", code)
```

```python
# map() - We can use any functionality on each element of the iterator insted of repeating it.

lst = [1,10,25,64,77,55]        # -> Defined a list
 
def into_2(element):            # -> Define a fucntion
    return element*2
 
result = list(map(into_2, lst)) # -> call nigesh pointing to same function (we can also call functions directly)
printer = print
printer(result)
```

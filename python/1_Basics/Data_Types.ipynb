##### COLLECTIONS MODULE

```python
# COUNTER : Counter is just to count frequency of elements in iterables and map it into dict.
from collections import Counter
logs = ["200", "500", "200", "404", "500", "500"]
error_count = Counter(logs)
print(error_count["500"])   # how many server errors
print(error_count.most_common(1))  # most frequent status
# so counter is all about the counting frequency of elelemtns in iterables and performing math and mutable operation on them.
# REAL CASE USE : If you are building log of something use this to count frequency for faster and lower memory.
```

```python
# DEFAULTDICT : for grouping logs and making data
from collections import defaultdict
logs = [("user1", "login"), ("user2", "view"), ("user1", "logout")]
user_actions = defaultdict(list)
for user, action in logs:
    user_actions[user].append(action)
print(user_actions)
# {'user1': ['login', 'logout'], 'user2': ['view']}
```

```python
# DEQUE : Itis used to optimise opration of list removal & adding of elements.
# deque = list but optimized for adding/removing from both ends (O(1))
from collections import deque
recent = deque(maxlen=3)
recent.append("req1")
recent.append("req2")
recent.append("req3")
recent.append("req4")  # auto removes "req1"

print(recent)
# ['req2', 'req3', 'req4']
```

```python
# NAMETUPLE : It is just a tuple with name like dictionary
from collections import namedtuple
User = namedtuple("User_name", ["name", "age"])
u = User("Pranit", 25)
print(u.name)  # clear
print(u.age)
```

Here’s the clean, practical rule-of-thumb:

- Counter → use when counting frequency
- Example: log errors, word counts, top-K items
- defaultdict → use when grouping or avoiding key errors
- Example: grouping logs by service → dict[list], dict[int]
- namedtuple → use when you want lightweight, readable objects
- Example: (x, y) → Point(x=1, y=2) instead of index-based tuples
- deque → use when fast insert/remove from both ends (O(1))
- Example: sliding window, queues, real-time streams

If you're building real systems (like your log anomaly detector), the combo is usually:
deque (window) + defaultdict (group) + Counter (count).

```python
# 1. ---- Find the most purchased product and its count ----
products = ["Pen", "Book", "Pen", "Bag", "Book", "Pen", "Bag"]

from collections import Counter
print(Counter(products).most_common(1))


# 2. ---- Group employees by department ----
employees = [
    ("IT", "Rahul"),
    ("HR", "Priya"),
    ("IT", "Amit"),
    ("Sales", "Neha"),
    ("HR", "Ravi")
]

from collections import defaultdict
results = defaultdict(list)
for i,a in employees:
    results[i].append(a)
print(results)


# 3. ---- A new customer "D" arrives, then the first two customers are served. ----
# Add "D" to the queue.
# Serve the first two customers.
from collections import deque
customers = deque(["A", "B", "C"])

customers.append("D")
customers.popleft()
customers.popleft()
print(customers)


# 4. ----Create a namedtuple called Employee and create this employee: ----
ID = 101
Name = "Rahul"
Salary = 50000

from collections import namedtuple
wusers = namedtuple("User", ["ID" , "Name", "Salary"])
users =  [ wusers(101 , "Rahul" , 50000), 
          wusers(102 , "Pranit" , 500000) ]
for user in users:
    print(user.ID, user.Name, user.Salary)
```

##### LINEAR ALGEBRA
Linear algebra is a branch of maths that focuses on study of vectors, matrics, tensors & linear transformation.

- It is used in neural networks, regression models and dim reduction algorithms.
- Rotation scaling, 3D production of the game are the matrix transformation.
- Working with high dim datasets and feature transformation requires fluency in linear algebra.

CONCEPTS :
    1. Scalar - single numerical value
    2. Vector - A list of values
    3. Matrix - 2D data 
    4. Tensor - 3D data.

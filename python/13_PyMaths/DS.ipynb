##### MEASUREMENT OF CENTRAL TENDENCY

```python
# Calculate Mean, Median, Mode
import numpy as np
from scipy import stats
arr = [3,4,6,8,5,3,1,3,5,56,780,5,34,34,56]

# -- Mean --
average  = sum(arr)/len(arr)
avg_2    = np.average(arr)
avg_3    = stats.gmean(arr)


# -- Median --
Med_1 = np.median(arr)


# -- Mode --
mode_1 = stats.mode(arr)
```

##### MEASUREMENT OF VARIABILITY

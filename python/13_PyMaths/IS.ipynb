##### HYPOTHESIS TESTING


```python
# ---- HYPOTHESIS TESTING ----
# Step 1 : Define Hypothesis
"""
 Does the numbers of hours have any effect on marks?
H0 - There is no relation
H1 - x  
"""

import pandas as pd
from scipy import stats
import numpy as np


df = pd.read_excel(r"C:\Users\PANRIT\OneDrive\Desktop\pca_student_data.xlsx")
a = 0.05    #-> Significance level
data = df[["Marks", "Hours"]].dropna()

#Pearson corelation hypothesis testing
r, p_value = stats.pearsonr(
    data["Hours"],
    data["Marks"]
)

alpha = 0.05

if p_value <= alpha:
    print("Rejected  H0", )
else :
    print("Not enough evidence to reject H0")
```

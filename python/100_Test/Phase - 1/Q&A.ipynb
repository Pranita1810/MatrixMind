##### COLLECTION MODULE
Q3. Explain the purpose of counter, defaultdict, deque, namedtuple and finally explain usage of each in real industry.
- Counter : It is used to count occurance of elemens in sequencial data. We can use them as counter like request error vs success counter to analyse the perforance of server by looking at numbers only
- deque   : deque allows user to append and pop values from list from both end which can be used for FIFO/LIFO method, temp memory storage.
- defaultdict : It avoids key error, if there is no key in dict and we try to add it will add normally which allow dynamic building and also helps grouping data based on keys.
- namedtuple  : namedtuple is dictionary with tuple.

```python
# CODING
# Q4. Guess the output:
from collections import Counter

data = ["SQL", "Python", "SQL", "Spark", "Python", "SQL", "Python"]

result = Counter(data)

print(result)
print(result.most_common(2))

# -> The output of this code is [(SQL, 3), (Python , 3)]
```

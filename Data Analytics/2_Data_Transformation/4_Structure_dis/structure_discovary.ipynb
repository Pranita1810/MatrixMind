#### -- 1. PCA (Principal Component Analysis) --
 - PCA is a dimentionality reduction technique and helps user to reduce the numbers of features in datasets while keeping the most 
   important information about the data. It converts complex data by transfoming correlated features into smaller set of uncorrelated components.

- It uses linear algebra to transform data into principal component.

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import pandas as pd

def run_pca(df, n_components=2):
    numeric_df = df.select_dtypes(include="number")

    scaler = StandardScaler()
    scaled = scaler.fit_transform(numeric_df)

    pca = PCA(n_components=n_components)    
    result = pca.fit_transform(scaled)

    pca_df = pd.DataFrame(
        result,
        columns=[f"PC{i+1}" for i in range(n_components)]
    )

    return pca_df, pca.explained_variance_ratio_
```

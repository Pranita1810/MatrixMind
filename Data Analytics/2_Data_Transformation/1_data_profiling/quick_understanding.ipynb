### DATA PROFILING
- Understanding the data is one of the most important key while analyzing the data. Traditionally we use pandas / polars 
  to describe get info about data but it does not give deep down analysis and understanding if we need deep down analysis we need
  to build our own pipeline whihc extract needed features and values to understand dat. 

- To solve this problem we have pre-built modules that allows user to pass df & parameters which will automatically create a short
  report about the data which helps user to understand data.

- These reports are not directly used in production level on industry level.

##### MODULES :
1. ydata-profiling  - Generates a full HTML report — distributions, correlations, missing values, duplicates
2. sweetviz         - Visually compares two dataframes side by side
3. dataprep         - Faster than ydata, cleaner UI, also does data cleanin
4. autoviz          - Auto-picks chart types based on column dtypes
5. dtale            - Launches a UI in browser, good for filtering/sorting live

####  -- 1. ydata-profiling --

```python
# -- Full HTML report --
import pandas as pd
from ydata_profiling import ProfileReport

df = pd.read_csv(r"C:\Users\PANRIT\ALL\Pranit Main\Study\Data Analytics\dirty_cafe_sales.csv", encoding='latin1')
profile = ProfileReport(df, title="My Report", explorative=True)
profile.to_file("report.html")  # open in browser
```
